```python
def loan_health(repayment_rate,status):
    status = status.lower()
    if status== "closed":
        return "Fully_Recovered"
    elif status == "defaulted":
        return "Defaulted-Critical"
    elif status == "active" and repayment_rate > 50:
        return "Good standing"
    elif status == "active" and repayment_rate<50:
        return "At Risk"
    else:
        return "end"
print(loan_health(100,"Closed"))
print(loan_health(15,"Defaulted"))
print(loan_health(75,"Active"))
print(loan_health(20,"Active"))

    
```

```python
def classify_npl(npl_ratio):
    if npl_ratio<5:
        return "healthy"
    if npl_ratio<=10:
        return "watch"
    else:
        return "crtical"
```

```python
loan_ids =[1,2,3,4,5]
customers = ["james kamau","esther akinyi","safaricom","wanjiru","mombasa spices"]
principals = [50000,250000,2500000,75000,900000]
repaid =[50000,150000,500000,15000,900000]
status =["closed","active","defaulted","active","closed"]

def classify_loan_risk(repayment_rate,status):
    if status == "closed":
        return "recovered"
    elif status == "defaulted":
        return "loss"
    elif repayment_rate >= 80 :
        return "low risk"
    elif repayment_rate >= 50 :
        return "medium risk"
    else:
        return "high risk"
for i in range(len(loan_ids)):
    rate = (repaid[i]/principals[i])*100
    risk = (classify_loan_risk(rate,status[i]))
    print(f"loan{loan_ids[i]}:{customers[i]}|repaid:{rate:.1f}|risk:{risk}")
total_principal = sum(principals)
total_repaid = sum(repaid)
portfolio_rate = (total_repaid/total_principal)*100
print(f"\nportfolio:kes{total_principal:,}|repaid:{portfolio_rate:.1f}%")

```

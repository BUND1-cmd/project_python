```python
def calculate_repayment_rate(total_repaid,principal):
    repaid =(total_repaid/principal)*100
    return round(repaid,2)
result1=calculate_repayment_rate(180000,1005000)
result2=calculate_repayment_rate(600000,3200000)
print(f"repayment_rate:{result1}")
print(f"repayment_rate:{result2}")
```

```python
def classify_salary(salary):
    if salary > 300000:
        return "senior"
    elif salary > 100000:
        return "mid"
    else:
        return"junior"
print(classify_salary(500000))
print(classify_salary(150000))
print(classify_salary(60000))
```

```python
def loan_health(repayment_rate,status):
    status = status.lower()
    if status== "closed":
        return "Fully_Recovered"
    elif status == "defaulted":
        return "Defaulted-Critical"
    elif status == "active" and repayment_rate > 50:
        return "Good standing"
    elif status == "active" and repayment_rate<50:
        return "At Risk"
    else:
        return "end"
print(loan_health(100,"Closed"))
print(loan_health(15,"Defaulted"))
print(loan_health(75,"Active"))
print(loan_health(20,"Active"))

    
```

```python
def classify_npl(npl_ratio):
    if npl_ratio<5:
        return "healthy"
    if npl_ratio<=10:
        return "watch"
    else:
        return "crtical"
```

```python
npl_values = [5.13,4.42,8.59,12.03,13.57]

for npl in npl_values:
    print(f"npl:{npl}%-{classify_npl(npl)}")
```

```python
years = [2020, 2021, 2022, 2023, 2024]
npl_values = [11.88, 10.89, 10.75, 12.34, 13.57]

for i,npl in enumerate(npl_values):
    print(f"year:{years[i]}|npl{npl}%|status:{classify_npl(npl)}")
```
